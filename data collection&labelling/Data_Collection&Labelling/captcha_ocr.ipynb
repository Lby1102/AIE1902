```python
import ddddocr, os, threading
```

```python
# Initialize OCR model (disable ad display)
ocr = ddddocr.DdddOcr(show_ad=False)
```

```python
def label_captcha(img_path, label_path):
    """
    Use ddddocr to recognize captcha and save the result to a text file.
    """
    if os.path.exists(label_path):
        return None  # Skip if already labeled
    
    with open(img_path, 'rb') as f:
        img = f.read()
    result = ocr.classification(img)
    with open(label_path, 'w', encoding='utf-8') as f:
        f.write(result)
```

```python
def process(start, end):
    """
    Process captcha images in batch from start to end index.
    """
    for idx in range(start, end):
        img_path = os.path.join("data_collection", f"img{idx+1}.png")
        label_path = os.path.join("data_label_ocr", f"img{idx+1}.txt")
        label_captcha(img_path, label_path)
```

```python
# Number of worker threads
worker_flow = 5
```

```python
# Create and start worker threads
thread = []
for worker in range(worker_flow):
    start = worker * 1000
    end = (worker + 1) * 1000
    t = threading.Thread(target=process, args=(start, end))
    thread.append(t)
    t.start()

for t in thread:
    t.join()
```
